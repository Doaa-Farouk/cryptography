### Simple DES Encryption and Decryption 
### ECB Mode

```python
from Crypto.Cipher import DES
from Crypto.Util.Padding import pad, unpad

key = b"mykey353"
# Key Must Be 8 bytes

plain_text = b"my secret text12"
# Plain Text Must Be a multi of 8 in ECB Mode

encrypted_text = DES.new(key, DES.MODE_ECB).encrypt(plain_text)
print(encrypted_text)
print(type(encrypted_text))
```

```python
# d = b"y\xcf\xbb\xbe\x18\xe9\x0f\xfe\xd1\xf8\x0e\x13\xcf\xa1~\xe9"
decrypted_text = DES.new(key, DES.MODE_ECB).decrypt(encrypted_text)
print(decrypted_text)
```
